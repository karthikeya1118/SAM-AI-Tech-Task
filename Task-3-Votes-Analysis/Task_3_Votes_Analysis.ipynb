```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_excel('/content/Data_Analyst_Internship_Restaurant_Dataset (1).xlsx')
print(df.head())
```

```python
# Restaurant with the highest votes
highest_votes = df.loc[df["Votes"].idxmax()]

print("Restaurant with Highest Votes:")
print("Restaurant Name:", highest_votes["Restaurant Name"])
print("City:", highest_votes["City"])
print("Votes:", highest_votes["Votes"])
print("Rating:", highest_votes["Rating"])


# Restaurant with the lowest votes
lowest_votes = df.loc[df["Votes"].idxmin()]

print("\nRestaurant with Lowest Votes:")
print("Restaurant Name:", lowest_votes["Restaurant Name"])
print("City:", lowest_votes["City"])
print("Votes:", lowest_votes["Votes"])
print("Rating:", lowest_votes["Rating"])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.scatter(df["Votes"], df["Rating"])

plt.title("Relationship Between Votes and Restaurant Ratings")
plt.xlabel("Number of Votes")
plt.ylabel("Restaurant Rating")

plt.tight_layout()
plt.show()
```

```python
correlation = df["Votes"].corr(df["Rating"])

print("Correlation between Votes and Rating:", round(correlation, 2))
```
