```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_excel('/content/Data_Analyst_Internship_Restaurant_Dataset (1).xlsx')
print(df.head())

```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Rating"], bins=10, edgecolor="black")

plt.title("Restaurant Ratings Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Restaurants")

plt.tight_layout()
plt.show()
```

```python
city_counts = df["City"].value_counts().head(10)

plt.figure(figsize=(10, 6))

city_counts.sort_values().plot(kind="barh")

plt.title("Top 10 Cities by Number of Restaurants")
plt.xlabel("Number of Restaurants")
plt.ylabel("City")

plt.tight_layout()
plt.show()
```

```python
cuisine_counts = {}

for cuisines in df["Cuisines"]:
    for cuisine in cuisines.split(","):
        cuisine = cuisine.strip()
        cuisine_counts[cuisine] = cuisine_counts.get(cuisine, 0) + 1

top_cuisines = pd.Series(cuisine_counts).sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 6))

top_cuisines.sort_values().plot(kind="barh")

plt.title("Top 10 Most Popular Cuisines")
plt.xlabel("Number of Restaurants")
plt.ylabel("Cuisine")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(df["Votes"], df["Rating"])

plt.title("Votes vs Restaurant Rating")
plt.xlabel("Number of Votes")
plt.ylabel("Restaurant Rating")

plt.tight_layout()
plt.show()
```
